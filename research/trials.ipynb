```python
# Ensure annotation demo

def hello(x:int, y:int) -> int:
    return x*y
```

```python
print(hello(2, 3))
```

```python
print(hello(2, "3")) # this is the problem we face so that's why we need ensure annotations decorator
```

```python
from ensure import ensure_annotations
```

```python
@ensure_annotations
def hello(x:int, y:int) -> int:
    return x*y
```

```python
print(hello(2, 3))
```

```python
print(hello(2, "3"))
```
